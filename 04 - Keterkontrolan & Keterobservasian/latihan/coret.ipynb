## 🧭 Peta Konsep Utama Topik 04
1. **Keterkontrolan Keadaan (*Controllability*):**
   - Matriks Kalman: $M_c = \begin{bmatrix} B & AB & A^2B & \dots & A^{n-1}B \end{bmatrix}$, syarat: $\operatorname{rank}(M_c) = n$.
   - Uji PBH: $\operatorname{rank}\begin{bmatrix} \lambda_i I - A & B \end{bmatrix} = n, \; \forall \lambda_i \in \sigma(A)$.
   - Uji Gilbert: Tidak ada baris nol pada $\tilde{B} = P^{-1} B$ untuk bentuk diagonal.
2. **Keterobservasian Keadaan (*Observability*):**
   - Matriks Kalman: $M_o = \begin{bmatrix} C \\ CA \\ CA^2 \\ \vdots \\ CA^{n-1} \end{bmatrix}$, syarat: $\operatorname{rank}(M_o) = n$.
   - Uji PBH: $\operatorname{rank}\begin{bmatrix} \lambda_i I - A \\ C \end{bmatrix} = n, \; \forall \lambda_i \in \sigma(A)$.
   - Uji Gilbert: Tidak ada kolom nol pada $\tilde{C} = C P$ untuk bentuk diagonal.
3. **Prinsip Dualitas Kalman:**
   - Sistem $(A, B)$ terkontrol $\iff$ Sistem $(A^T, B^T)$ terobservasi.
4. **Transformasi Koordinat ke Bentuk Kanonik:**
   - Matriks $P$ untuk CCF dan Matriks $Q$ untuk OCF.
5. **Dekomposisi Struktural Kalman:**
   - Pemisahan 4 subruang: $x_{co}, x_{c\bar{o}}, x_{\bar{c}o}, x_{\bar{c}\bar{o}}$.


```
# 🛠️ Toolkit Simbolik Pengujian Keterkontrolan & Keterobservasian dengan SymPy
import sympy as sp
sp.init_printing()

def uji_keterkontrolan(A, B):
    """Menghitung matriks keterkontrolan Kalman dan rank-nya."""
    n = A.shape[0]
    cols = [B]
    cur = B
    for _ in range(1, n):
        cur = A * cur
        cols.append(cur)
    Mc = sp.Matrix.hstack(*cols)
    rank = Mc.rank()
    is_controllable = (rank == n)
    print(f"--- Uji Keterkontrolan Kalman (n = {n}) ---")
    print("Matriks Mc:")
    display(Mc)
    print(f"Rank(Mc) = {rank} -> {'TERKONTROL PENUH' if is_controllable else 'TIDAK TERKONTROL'}")
    if Mc.is_square:
        print(f"Det(Mc)  = {Mc.det()}")
    return Mc

def uji_keterobservasian(A, C):
    """Menghitung matriks keterobservasian Kalman dan rank-nya."""
    n = A.shape[0]
    rows = [C]
    cur = C
    for _ in range(1, n):
        cur = cur * A
        rows.append(cur)
    Mo = sp.Matrix.vstack(*rows)
    rank = Mo.rank()
    is_observable = (rank == n)
    print(f"\n--- Uji Keterobservasian Kalman (n = {n}) ---")
    print("Matriks Mo:")
    display(Mo)
    print(f"Rank(Mo) = {rank} -> {'TEROBSERVASI PENUH' if is_observable else 'TIDAK TEROBSERVASI'}")
    if Mo.is_square:
        print(f"Det(Mo)  = {Mo.det()}")
    return Mo

print("Toolkit Pengujian Keterkontrolan & Keterobservasian siap digunakan!")
```

```
# Verifikasi Soal 1
A1 = sp.Matrix([
    [1, 2, 0],
    [0, -1, 1],
    [0, 0, -2]
])
B1 = sp.Matrix([
    [1],
    [0],
    [1]
])

Mc1 = uji_keterkontrolan(A1, B1)
```

## 🔍 Pembahasan Soal 2(b): Kenapa Rank Matriks Bisa Dilihat dari Minor $2 \times 2$ vs OBE?

Dalam Aljabar Linear, terdapat **dua metode yang setara (ekuivalen)** untuk menentukan rank suatu matriks:

### 1. Metode Operasi Baris Elementer (OBE / Eliminasi Gauss)
- **Prinsip:** Mengubah matriks ke bentuk eselon baris (*Row Echelon Form / REF*).
- **Definisi Rank:** Jumlah baris tak-nol (jumlah satu utama / pivot).
- **Pengerjaan Soal 2:**
  $$M_o = \begin{bmatrix} 2 & 1 & 0 \\ 0 & 2 & 1 \\ -6 & -11 & -4 \end{bmatrix}$$
  - Baris 3 dieliminasi dengan $B_3 \leftarrow B_3 + 3B_1$:
    $$\begin{bmatrix} 2 & 1 & 0 \\ 0 & 2 & 1 \\ 0 & -8 & -4 \end{bmatrix}$$
  - Baris 3 dieliminasi dengan $B_3 \leftarrow B_3 + 4B_2$:
    $$\begin{bmatrix} 2 & 1 & 0 \\ 0 & 2 & 1 \\ 0 & 0 & 0 \end{bmatrix}$$
  - Terdapat **2 baris tak-nol** (baris 1 dan baris 2), sedangkan baris 3 menjadi baris nol $[0, 0, 0]$.
  - Kesimpulan: $\operatorname{rank}(M_o) = 2$.

---

### 2. Metode Determinan / Minor Terbesar (*Determinantal Rank*)
- **Definisi Teori (Teorema Kronecker):** 
  *Rank suatu matriks adalah ukuran terbesar $k$ dari submatriks persegi $k \times k$ yang memiliki determinan $\ne 0$.*
- **Alur Logika:**
  1. Cek ukuran $3 \times 3$: $\det(M_o) = 0 \implies$ Matriks tidak ber-rank 3, sehingga $\operatorname{rank}(M_o) \le 2$.
  2. Cek ukuran $2 \times 2$: Cukup cari **SATU** saja submatriks $2 \times 2$ yang determinannya tak-nol.
     Pilih submatriks di pojok kiri atas:
     $$\det \begin{bmatrix} 2 & 1 \\ 0 & 2 \end{bmatrix} = (2)(2) - (1)(0) = 4 \ne 0$$
  3. Karena ditemukan minor $2 \times 2$ dengan determinan $\ne 0$, maka terjamin $\operatorname{rank}(M_o) \ge 2$.
  4. Karena $\operatorname{rank}(M_o) \le 2$ dan $\operatorname{rank}(M_o) \ge 2$, maka **pasti $\operatorname{rank}(M_o) = 2$**!

**Kapan menggunakan Minor vs OBE?**
- **Metode Minor** sangat cepat dan disukai saat ujian atau perhitungan manual pada matriks ukuran kecil ($2 \times 2$ atau $3 \times 3$), atau ketika ada variabel simbolik ($\alpha, \beta$) karena menghindari pembagian pecahan aljabar.
- **Metode OBE** sangat sistematis bila ukuran matriks besar ($4 \times 4$ ke atas) atau bila kita ingin mencari basis ruang nol (kernel/nullspace).

```
# Verifikasi Soal 2 & Perbandingan Hasil OBE vs Minor
A2 = sp.Matrix([
    [0, 1, 0],
    [0, 0, 1],
    [-6, -11, -6]
])
C2 = sp.Matrix([[2, 1, 0]])

Mo2 = uji_keterobservasian(A2, C2)

print("\n=== PEMBUKTIAN DUA CARA MENGHITUNG RANK ===")
# Cara 1: Bentuk Eselon Baris (RREF / OBE)
rref_matrix, pivots = Mo2.rref()
print("1. Melalui OBE / Bentuk Eselon Baris Tereduksi (RREF):")
display(rref_matrix)
print(f"   Jumlah baris tak-nol (pivot) = {len(pivots)} -> Rank = {len(pivots)}")

# Cara 2: Minor 2x2
sub_matrix = Mo2[0:2, 0:2]
print("\n2. Melalui Minor 2x2 (Pojok Kiri Atas):")
display(sub_matrix)
print(f"   Det(minor 2x2) = {sub_matrix.det()} != 0 -> Terbukti Rank = 2")
```
