Bagian $G_A$ pada penyelesaian Nomor 3 merupakan hasil penyederhanaan loop umpan balik (feedback loop) internal pertama yang ada di sebelah kiri diagram.

Berikut adalah penjelasan detail dan asal-usul perhitungannya:

1. Letak Bagian $G_A$ pada Diagram Blok
Jika kita perhatikan bagian kiri diagram:

Sinyal melewati dua blok berurutan (kaskade/seri): $G_1$ dan $G_2$, sehingga fungsi transfer jalurnya adalah: $$G_{\text{maju}} = G_1 G_2$$
Tepat setelah blok $G_2$, ada titik percabangan (pick-off point) yang mengambil sinyal keluaran tersebut dan mengumpankannya kembali melalui blok $H_1$.
Sinyal dari $H_1$ masuk ke titik penjumlahan (summing point) dengan tanda minus ($-$), yang berarti ini adalah umpan balik negatif (negative feedback).

2. Penurunan Matematis Langkah demi Langkah
Misalkan:

$E(s)$ = sinyal masukan yang masuk ke titik penjumlahan sebelum blok $G_1$

$X_A(s)$ = sinyal keluaran tepat setelah blok $G_2$
Dari diagram blok: Sinyal yang keluar dari titik penjumlahan adalah selisih antara masukan $E(s)$ dan sinyal umpan balik $H_1 X_A(s)$: $$\text{Sinyal setelah titik penjumlahan} = E(s) - H_1 X_A(s)$$

Sinyal tersebut kemudian dikalikan dengan blok $G_1$ dan $G_2$ untuk menghasilkan $X_A(s)$: $$X_A(s) = G_1 G_2 \left[ E(s) - H_1 X_A(s) \right]$$

Jabarkan ruas kanan: $$X_A(s) = G_1 G_2 E(s) - G_1 G_2 H_1 X_A(s)$$

Pindahkan suku yang mengandung $X_A(s)$ ke ruas kiri: $$X_A(s) + G_1 G_2 H_1 X_A(s) = G_1 G_2 E(s)$$

Faktorkan $X_A(s)$: $$(1 + G_1 G_2 H_1) X_A(s) = G_1 G_2 E(s)$$

Bagi kedua ruas dengan $(1 + G_1 G_2 H_1)$ untuk mendapatkan rasio keluaran terhadap masukan: $$\frac{X_A(s)}{E(s)} = \frac{G_1 G_2}{1 + G_1 G_2 H_1}$$

Rasio inilah yang didefinisikan sebagai satu blok pengganti $G_A$: $$G_A = \frac{G_1 G_2}{1 + G_1 G_2 H_1}$$

3. Menggunakan Rumus Cepat Blok Umpan Balik (Feedback Rule)
Secara umum dalam Teori Kontrol, sebuah loop tertutup dengan:

Jalur maju = $G$
Jalur umpan balik = $H$ (dengan tanda minus / negatif)
dapat langsung disederhanakan menjadi satu blok tunggal: $$\text{Fungsi Transfer} = \frac{G}{1 + GH}$$

Karena pada subsistem ini jalur majunya adalah $G = G_1 G_2$ dan jalur baliknya adalah $H = H_1$, maka langsung diperoleh: $$G_A = \frac{G_1 G_2}{1 + G_1 G_2 H_1}$$
